```python
import threading

def timed_input(prompt, timeout=5):
    result = {"value": None}

    def ask():
        try:
            result["value"] = input(prompt)
        except EOFError:
            result["value"] = None

    thread = threading.Thread(target=ask)
    thread.daemon = True
    thread.start()
    thread.join(timeout)

    if thread.is_alive():
        return None  # timeout
    return result["value"]

def note_val(timeout=5):
    val = timed_input("Enter MIDI note number (0-127) or -1 to exit: ", timeout)
    if val is None:
        return None  # no input given in time
    try:
        return int(val)
    except ValueError:
        return None
```

```python
note = 1
velocity = 64
channel = 0
duration = 0.5

while True:
    if note == -1: break
    val = note_val(timeout=5)
    if val is not None and 0 <= val <= 127: note = val
    with mido.open_output(mido.get_output_names()[1]) as outport:
        outport.send(mido.Message('note_on', note=note, velocity=velocity, channel=channel))
        time.sleep(duration)
        outport.send(mido.Message('note_off', note=note, velocity=0, channel=channel))
```
